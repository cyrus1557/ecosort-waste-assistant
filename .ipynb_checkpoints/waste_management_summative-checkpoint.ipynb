## Overview

You are a data scientist at "EcoSort," a technology company that specializes in developing AI solutions for waste management. EcoSort has partnered with Metro City's waste management department to develop an intelligent waste management assistant that can help residents properly dispose of waste items so less time is spent sorting material at facilities.

This assistant needs to:

1. Identify waste materials from images uploaded by residents (CNN)
2. Classify waste items based on text descriptions provided by residents (RNN/Transformer)
3. Generate specific recycling instructions based on identified waste type and city policies (Generative Transformer with RAG)

Your task is to build this integrated system using the RealWaste dataset along with generated text data that simulates real-world waste management operations.

## How to run this notebook

- Put realwaste.zip, waste_descriptions.csv and waste_policy_documents.json in the Drive folder set as DRIVE_DIR in the first code cell.
- Use a T4 GPU and run top to bottom (about 40 minutes; the Part 2 tuning experiments are the slowest part).
- Trained models are saved to the models folder on Drive (CNN, DistilBERT, flan-t5).
- Pipeline: image → CNN or text → DistilBERT, then category → retrieval + flan-t5 → instructions with sources (Part 5).

### 1.3 Create Data Pipelines

```python
# Run this code to setup the images properly into train, validation, and test sets
# Set your data directory path - update this with your actual path
import pathlib
data_dir = pathlib.Path('RealWaste')

# Parameters
BATCH_SIZE = 32
IMG_HEIGHT = 224
IMG_WIDTH = 224

# Calculate the total number of classes automatically from the directory structure
num_classes = len([item for item in data_dir.glob('*') if item.is_dir()])
print(f"Number of classes: {num_classes}")

# List all class folders
class_names = sorted([item.name for item in data_dir.glob('*') if item.is_dir()])
print(f"Class names: {class_names}")

# Count all images
image_count = len(list(data_dir.glob('*/*.jpg'))) + len(list(data_dir.glob('*/*.png')))
print(f"Total images found: {image_count}")

# Create a dataset using tf.keras.utils.image_dataset_from_directory
# This will automatically split the data into training and validation sets
train_ds = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=0.2,  # 20% for validation
    subset="training",
    seed=42,
    image_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE,
    label_mode='categorical',  # For one-hot encoded labels
    shuffle=True
)

validation_ds = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=0.2,  # 20% for validation
    subset="validation",
    seed=42,
    image_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE,
    label_mode='categorical',  # For one-hot encoded labels
    shuffle=True
)

# Create a separate test dataset by taking part of the validation set
# First, let's get the number of batches in the validation set
# NOTE: original take/skip lines commented out. validation_ds reshuffles every time it is read,
# so take() and skip() drew from different orders and 129 of the 480 test images were also
# in validation (hash check in the next cell). splitting the held-out files by path instead,
# so validation and test are fixed and never share an image
# val_batches = tf.data.experimental.cardinality(validation_ds)
# test_dataset = validation_ds.take(val_batches // 2)
# validation_ds = validation_ds.skip(val_batches // 2)

from sklearn.model_selection import train_test_split

# the 950 held-out files (seed=42 fixes which files these are, same as before)
holdout_files = np.array(validation_ds.file_paths)
holdout_labels = np.array([class_names.index(pathlib.Path(p).parent.name) for p in holdout_files])

# half validation, half test, stratified so both sets have the same class mix
val_files, test_files, val_labels, test_labels = train_test_split(
    holdout_files, holdout_labels, test_size=0.5, stratify=holdout_labels, random_state=42)

# make sure no file is shared between any two sets
train_files = set(train_ds.file_paths)
assert not (set(val_files) & set(test_files)), "validation and test overlap"
assert not (train_files & set(val_files)), "train and validation overlap"
assert not (train_files & set(test_files)), "train and test overlap"

# load and resize the same way image_dataset_from_directory does (bilinear, float 0-255)
def load_from_path(path, label):
    img = tf.io.read_file(path)
    img = tf.image.decode_image(img, channels=3, expand_animations=False)
    img = tf.image.resize(img, (IMG_HEIGHT, IMG_WIDTH), method='bilinear')
    return img, tf.one_hot(label, num_classes)

def make_ds(files, labels):
    ds = tf.data.Dataset.from_tensor_slices((files, labels))
    ds = ds.map(load_from_path, num_parallel_calls=tf.data.AUTOTUNE)
    return ds.batch(BATCH_SIZE)   # no shuffling, so the order is the same every time

validation_ds = make_ds(val_files, val_labels)
test_dataset = make_ds(test_files, test_labels)

print(f"Number of training batches: {tf.data.experimental.cardinality(train_ds)}")
print(f"Number of validation batches: {tf.data.experimental.cardinality(validation_ds)}")
print(f"Number of test batches: {tf.data.experimental.cardinality(test_dataset)}")
print(f"Images: train {len(train_files)}, validation {len(val_files)}, test {len(test_files)}")

# Configure dataset for performance
AUTOTUNE = tf.data.AUTOTUNE

train_ds = train_ds.cache().prefetch(buffer_size=AUTOTUNE)
validation_ds = validation_ds.cache().prefetch(buffer_size=AUTOTUNE)
test_dataset = test_dataset.cache().prefetch(buffer_size=AUTOTUNE)
```

**Note on the image split**

- The provided take()/skip() code drew validation and test from a set that reshuffles on every read. A hash check found 129 shared images.
- Fixed by splitting the 950 held-out files by file path (stratified): 475 validation and 475 test, no shared images, same files on every run. The training set is unchanged.
- The file paths also give real test images for Part 5.
- The stage 1, first fine-tuning and tuning-experiment results in Part 2 were produced before this fix, so their test figures may be slightly optimistic. The final CNN was retrained and evaluated on the fixed split.

```python
# TODO: Create a text preprocessing pipeline
# - Tokenization
# - Text cleaning
# - Split data into train and test
# - Create embeddings/features


import re
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.feature_extraction.text import TfidfVectorizer

# first look at what the punctuation/numbers actually are (904 rows had them)
odd_chars = desc_df['description'].str.findall(r'[^a-zA-Z\s]').explode().dropna()
print("Non-letter characters found:", Counter(odd_chars).most_common(10))
print(desc_df[desc_df['description'].str.contains(r'[^a-zA-Z\s]')]['description'].head(5).tolist())

# cleaning function: lowercase, remove apostrophes, turn other symbols into spaces
def clean_text(text):
    text = text.lower()
    text = text.replace("'", "")
    text = re.sub(r'[^a-z\s]', ' ', text)   # hyphens, numbers etc become spaces
    text = re.sub(r'\s+', ' ', text).strip()  # fix the extra spaces
    return text

desc_df['clean_text'] = desc_df['description'].apply(clean_text)

# check a few before and after
print("\nBefore vs after cleaning:")
print(desc_df[['description', 'clean_text']].sample(5, random_state=1))

# tokenization (just splitting on spaces for now, the bert tokenizer comes in part 3)
desc_df['tokens'] = desc_df['clean_text'].str.split()

# duplicates and label conflicts again, now on the cleaned text
print("\nDuplicates after cleaning:", desc_df['clean_text'].duplicated().sum())
conflicts = desc_df.groupby('clean_text')['category'].nunique()
print("Texts with conflicting labels after cleaning:", (conflicts > 1).sum())

# remove conflicting texts completely, then drop the duplicates
conflict_texts = conflicts[conflicts > 1].index
text_df = desc_df[~desc_df['clean_text'].isin(conflict_texts)]
text_df = text_df.drop_duplicates(subset='clean_text').reset_index(drop=True)
print("Rows left:", len(text_df), "(from", len(desc_df), ")")

# encode the category names as numbers
le = LabelEncoder()
text_df['label'] = le.fit_transform(text_df['category'])
print("\nLabel mapping:", dict(zip(le.classes_, range(len(le.classes_)))))
# need the same class order as the image model so they work together later
print("Same order as image classes:", list(le.classes_) == class_names)

# split 70% train, 15% val, 15% test (stratified so every class is in each split)
train_df, temp_df = train_test_split(
    text_df, test_size=0.30, stratify=text_df['label'], random_state=42)
val_df, test_df = train_test_split(
    temp_df, test_size=0.50, stratify=temp_df['label'], random_state=42)
print("\nTrain / Val / Test sizes:", len(train_df), len(val_df), len(test_df))

# features: tf-idf with single words and word pairs, fit on train only
tfidf = TfidfVectorizer(ngram_range=(1, 2))
X_train = tfidf.fit_transform(train_df['clean_text'])
X_val = tfidf.transform(val_df['clean_text'])
X_test = tfidf.transform(test_df['clean_text'])
y_train, y_val, y_test = train_df['label'], val_df['label'], test_df['label']
print("TF-IDF shapes (train, val, test):", X_train.shape, X_val.shape, X_test.shape)

# how many words in the test set never appeared in training? (useful for error analysis)
train_vocab = set(w for toks in train_df['tokens'] for w in toks)
test_vocab = set(w for toks in test_df['tokens'] for w in toks)
unseen = test_vocab - train_vocab
print("Test words never seen in train:", len(unseen), list(unseen)[:10])
```

**Observations from text preprocessing**

- Cleaning lowercases the text and turns hyphens and slashes into spaces. After removing duplicates, 4,939 rows remain.
- Stratified 70/15/15 split (3,457 / 741 / 741). TF-IDF is fit on the training data only.
- Every test word also appears in training (closed vocabulary), a limitation of the generated data.

```python
# TODO: Prepare documents for RAG
# - Document preprocessing
# - Create embeddings for retrieval


all_categories = sorted(desc_df['category'].unique())
rag_docs = []

# helper to split a document into sections using the ALL CAPS headings that end with ':'
def split_sections(text):
    sections = {}
    current = None
    for line in text.split('\n'):
        line = line.strip()
        if not line:
            continue
        if line.endswith(':') and line.upper() == line:
            current = line[:-1]
            sections[current] = []
        elif current is not None:
            sections[current].append(line)
    return sections

# --- 1. policy documents ---
# policies 1-9 cover one category, keeping each as ONE chunk (they are short and
# splitting would separate the acceptable items from the non-acceptable ones)
# policies 10-14 are combined, so splitting them by category heading
seen_general = False
for doc in policy_docs:
    header = f"{doc['policy_type']} (Metro City, effective {doc['effective_date']})"
    if len(doc['categories_covered']) == 1:
        rag_docs.append({
            'text': header + '\n' + doc['document_text'].strip(),
            'category': doc['categories_covered'][0],
            'source': 'category_policy',
            'policy_id': doc['policy_id'],
            'priority': 1
        })
    else:
        for heading, lines in split_sections(doc['document_text']).items():
            if heading == 'GENERAL REQUIREMENTS':
                # same general rules in all combined docs so only keep it once
                if seen_general:
                    continue
                seen_general = True
                cat = 'General'
            else:
                cat = heading.replace(' GUIDELINES', '').title()
            rag_docs.append({
                'text': header + '\n' + heading.title() + ':\n' + '\n'.join(lines),
                'category': cat,
                'source': 'combined_policy',
                'policy_id': doc['policy_id'],
                'priority': 2
            })

# --- 2. disposal instructions from the csv, one chunk per category ---
for cat in all_categories:
    sub = desc_df[desc_df['category'] == cat]
    steps = sub['disposal_instruction'].unique()
    confusion = sub['common_confusion'].dropna().unique()
    material = sub['material_composition'].unique()

    text = f"Disposal instructions for {cat}:\n" + '\n'.join('- ' + s for s in steps)
    if len(confusion) > 0:
        text += "\nCommon confusion: " + ' '.join(confusion)
    text += "\nMaterial: " + ' '.join(material)

    rag_docs.append({
        'text': text,
        'category': cat,
        'source': 'disposal_instructions',
        'policy_id': None,
        'priority': 1
    })

# put everything in a dataframe and give each chunk an id
rag_df = pd.DataFrame(rag_docs)
rag_df['chunk_id'] = ['chunk_' + str(i) for i in range(len(rag_df))]
rag_df['n_words'] = rag_df['text'].str.split().str.len()

# --- checks ---
print("Total chunks:", len(rag_df))
print("\nChunks per category and source:")
print(rag_df.groupby(['category', 'source']).size().unstack(fill_value=0))
print("\nChunk length in words:")
print(rag_df['n_words'].describe())

# look at one combined chunk and one disposal chunk to see they look right
print("\n--- combined glass chunk ---")
print(rag_df[(rag_df['source'] == 'combined_policy') & (rag_df['category'] == 'Glass')]['text'].iloc[0])
print("\n--- disposal chunk ---")
print(rag_df[rag_df['source'] == 'disposal_instructions']['text'].iloc[0])
```

**Observations from RAG document preparation**

- 37 chunks: 9 category policies kept whole, 18 category sections from the combined policies, 1 general chunk and 9 disposal-instruction chunks from the CSV.
- Each chunk is tagged with category, source and priority (category policy first) for filtered retrieval.
- Chunks are short (27 to 112 words), so no further splitting is needed.

## Part 2: Waste Material Classification with CNN

### 2.1 Preprocess Images

```python
# TODO: Implement image preprocessing
# - Apply the preprocessing pipeline created earlier


from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
from sklearn.utils.class_weight import compute_class_weight

# images are already resized to 224x224 in the pipeline cell, now need to normalize the pixels
# mobilenetv2 wants pixel values between -1 and 1, preprocess_input does that
def normalize(images, labels):
    return preprocess_input(images), labels

# making new datasets so rerunning this cell doesnt normalize twice
train_prep = train_ds.map(normalize, num_parallel_calls=AUTOTUNE).prefetch(AUTOTUNE)
val_prep = validation_ds.map(normalize, num_parallel_calls=AUTOTUNE).prefetch(AUTOTUNE)
test_prep = test_dataset.map(normalize, num_parallel_calls=AUTOTUNE).prefetch(AUTOTUNE)

# check the pixel range before and after on one batch
for images, labels in train_ds.take(1):
    print("Before:", images.shape, "min", float(tf.reduce_min(images)), "max", float(tf.reduce_max(images)))
    norm_images, _ = normalize(images, labels)
    print("After: min", float(tf.reduce_min(norm_images)), "max", float(tf.reduce_max(norm_images)))

# class weights to handle the imbalance (textile has 318 images, plastic 921)
# getting the class index of every training image first
train_labels = np.concatenate([np.argmax(y.numpy(), axis=1) for _, y in train_ds])
weights_arr = compute_class_weight('balanced', classes=np.arange(num_classes), y=train_labels)
class_weights = dict(enumerate(weights_arr))

weights_df = pd.DataFrame({
    'class': class_names,
    'train_images': np.bincount(train_labels, minlength=num_classes),
    'weight': weights_arr.round(2)
})
print("\nClass weights:")
print(weights_df)

# data augmentation, will be used inside the model so it only runs while training
data_augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomFlip('horizontal_and_vertical'),
    tf.keras.layers.RandomRotation(0.1),
    tf.keras.layers.RandomZoom(0.1),
], name='data_augmentation')

# helper for a single image (needed later for the integrated assistant)
# does the same resize + normalize as the training images
def prepare_image(img_path):
    img = tf.keras.utils.load_img(img_path, target_size=(IMG_HEIGHT, IMG_WIDTH))
    arr = tf.keras.utils.img_to_array(img)
    arr = np.expand_dims(arr, axis=0)
    return preprocess_input(arr)
```

**Observations from image preprocessing**

- Images are resized to 224x224 and scaled to -1 to 1, as MobileNetV2 expects.
- Class weights from the training set (Textile Trash 1.74, Plastic 0.57) handle the imbalance.
- Augmentation (flip, rotation, zoom) is part of the model and only runs during training.

### 2.2 Implement CNN Model with Transfer Learning

```python
# TODO: Select an appropriate base model and implement transfer learning
# - Choose from MobileNet, EfficientNet, etc.
# - Add custom classification layers for the 9 waste categories
# - Configure loss function and metrics


from tensorflow.keras import layers, models, regularizers

IMG_SHAPE = (IMG_HEIGHT, IMG_WIDTH, 3)

# function so i can change dropout, dense units and l2 later when tuning
def build_model(dropout_rate=0.3, dense_units=256, l2_reg=1e-4):
    # mobilenetv2 pretrained on imagenet, without its top classifier
    base_model = tf.keras.applications.MobileNetV2(
        input_shape=IMG_SHAPE, include_top=False, weights='imagenet')
    base_model.trainable = False   # freeze for now, only train my new layers first

    inputs = tf.keras.Input(shape=IMG_SHAPE)
    x = data_augmentation(inputs)              # only active while training
    x = base_model(x, training=False)          # keep batchnorm layers in inference mode
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dropout(dropout_rate)(x)
    x = layers.Dense(dense_units, activation='relu',
                     kernel_regularizer=regularizers.l2(l2_reg))(x)
    x = layers.Dropout(dropout_rate)(x)
    outputs = layers.Dense(num_classes, activation='softmax')(x)  # 9 waste categories

    model = models.Model(inputs, outputs)
    return model, base_model

# loss is categorical crossentropy because the labels are one-hot
def compile_model(model, learning_rate=1e-3):
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=learning_rate),
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )
    return model

model, base_model = build_model()
model = compile_model(model)
model.summary()

# how many parameters are actually being trained
trainable = sum(tf.size(w).numpy() for w in model.trainable_weights)
total = model.count_params()
print(f"Trainable params: {trainable:,} of {total:,}")
```

```python
# TODO: Evaluate model performance
# - Calculate accuracy on test set
# - Generate confusion matrix
# - Analyze error patterns

from sklearn.metrics import confusion_matrix, classification_report, accuracy_score

# function so i can run the same thing again after fine tuning and compare
def evaluate_cnn(model, dataset, title):
    # one pass over the test set, collecting labels and predictions from the same batches
    # so the order always matches
    y_true, y_pred = [], []
    wrong_imgs, wrong_true, wrong_pred, wrong_conf = [], [], [], []
    for images, labels in dataset:
        probs = model.predict(images, verbose=0)
        true_idx = np.argmax(labels.numpy(), axis=1)
        pred_idx = np.argmax(probs, axis=1)
        y_true.extend(true_idx)
        y_pred.extend(pred_idx)
        # keep the wrong ones so i can look at them later
        wrong = pred_idx != true_idx
        wrong_imgs.extend(images.numpy()[wrong])
        wrong_true.extend(true_idx[wrong])
        wrong_pred.extend(pred_idx[wrong])
        wrong_conf.extend(probs[wrong].max(axis=1))

    y_true, y_pred = np.array(y_true), np.array(y_pred)
    acc = accuracy_score(y_true, y_pred)
    print(f"{title}: test accuracy = {acc:.4f} on {len(y_true)} images\n")
    print(classification_report(y_true, y_pred, target_names=class_names, digits=3))

    # confusion matrix: counts on the left, percent of each true class on the right
    cm = confusion_matrix(y_true, y_pred)
    cm_norm = cm / cm.sum(axis=1, keepdims=True)
    fig, axes = plt.subplots(1, 2, figsize=(18, 7))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=class_names, yticklabels=class_names, ax=axes[0])
    axes[0].set_title(f'{title}: confusion matrix (counts)')
    sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='Blues',
                xticklabels=class_names, yticklabels=class_names, ax=axes[1])
    axes[1].set_title(f'{title}: confusion matrix (share of true class)')
    for ax in axes:
        ax.set_xlabel('Predicted')
        ax.set_ylabel('True')
        ax.tick_params(axis='x', rotation=45)
    plt.tight_layout()
    plt.show()

    # most common mistakes
    pairs = []
    for i in range(num_classes):
        for j in range(num_classes):
            if i != j and cm[i, j] > 0:
                pairs.append((cm[i, j], class_names[i], class_names[j]))
    pairs.sort(reverse=True)
    print("Most common mistakes (true -> predicted):")
    for n, t, p in pairs[:6]:
        print(f"  {t} -> {p}: {n}")

    return {'acc': acc, 'y_true': y_true, 'y_pred': y_pred, 'cm': cm,
            'wrong_imgs': wrong_imgs, 'wrong_true': wrong_true,
            'wrong_pred': wrong_pred, 'wrong_conf': wrong_conf}

# show some of the wrong predictions with the true label, predicted label and confidence
def show_errors(results, n=8):
    n = min(n, len(results['wrong_imgs']))
    idx = random.sample(range(len(results['wrong_imgs'])), n)
    fig, axes = plt.subplots(2, 4, figsize=(14, 7))
    for ax, k in zip(axes.flatten(), idx):
        img = (results['wrong_imgs'][k] + 1) / 2   # undo the -1 to 1 scaling to display
        ax.imshow(np.clip(img, 0, 1))
        ax.set_title(f"true: {class_names[results['wrong_true'][k]]}\n"
                     f"pred: {class_names[results['wrong_pred'][k]]} ({results['wrong_conf'][k]:.2f})",
                     fontsize=9)
        ax.axis('off')
    plt.tight_layout()
    plt.show()

stage1_results = evaluate_cnn(model, test_prep, 'MobileNetV2 stage 1 (frozen base)')
show_errors(stage1_results)
```

**Observations from the stage 1 evaluation**

- Test accuracy is about 0.83, close to validation.
- Strongest classes: Vegetation, Food Organics and Textile Trash. Weakest: Miscellaneous Trash (errors spread across all classes), Cardboard and Glass.
- Main confusions: Metal/Plastic, Plastic/Glass and Cardboard/Paper, which look alike in photos.
- Several errors are made with high confidence, so confidence alone is not a reliable error signal.

### 2.4 Fine-tune the Model

```python
# TODO: Tune model parameters to improve performance
# - Adjust learning rate
# - Add regularization, dropout
# - Modify architecture if needed

# fine tuning: unfreeze the top part of mobilenetv2 and train with a very small learning rate
# the early layers learn general stuff (edges, textures) so those stay frozen
def fine_tune(model, base_model, n_unfreeze=40, lr=5e-5, epochs=15, name='cnn_finetune'):
    # unfreeze the base, then freeze everything except the last n layers
    base_model.trainable = True
    for layer in base_model.layers[:-n_unfreeze]:
        layer.trainable = False
    # keep the batchnorm layers frozen (better when the dataset is small)
    for layer in base_model.layers:
        if isinstance(layer, tf.keras.layers.BatchNormalization):
            layer.trainable = False

    # have to compile again after changing trainable
    model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=lr),
                  loss='categorical_crossentropy', metrics=['accuracy'])
    trainable = sum(tf.size(w).numpy() for w in model.trainable_weights)
    print(f"{name}: unfroze last {n_unfreeze} layers, trainable params = {trainable:,}")

    cbs = [
        tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=4, restore_best_weights=True),
        tf.keras.callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=2, min_lr=1e-7),
        tf.keras.callbacks.ModelCheckpoint(f"{MODEL_DIR}/{name}.keras",
                                           monitor='val_loss', save_best_only=True)
    ]
    hist = model.fit(train_prep, validation_data=val_prep, epochs=epochs,
                     class_weight=class_weights, callbacks=cbs)
    return hist

# first run: last 40 layers, learning rate 5e-5
ft_history = fine_tune(model, base_model, n_unfreeze=40, lr=5e-5, epochs=15,
                       name='cnn_finetune_40')

# plot train vs validation again
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(ft_history.history['accuracy'], label='train')
axes[0].plot(ft_history.history['val_accuracy'], label='validation')
axes[0].set_title('Fine-tuning accuracy')
axes[0].set_xlabel('Epoch')
axes[0].legend()
axes[1].plot(ft_history.history['loss'], label='train')
axes[1].plot(ft_history.history['val_loss'], label='validation')
axes[1].set_title('Fine-tuning loss')
axes[1].set_xlabel('Epoch')
axes[1].legend()
plt.tight_layout()
plt.show()

# compare with stage 1 on the validation set (at each one's best epoch)
s1_best = int(np.argmin(history.history['val_loss']))
ft_best = int(np.argmin(ft_history.history['val_loss']))
print(f"Stage 1     -> val accuracy {history.history['val_accuracy'][s1_best]:.3f}, "
      f"val loss {history.history['val_loss'][s1_best]:.3f}")
print(f"Fine-tuned  -> val accuracy {ft_history.history['val_accuracy'][ft_best]:.3f}, "
      f"val loss {ft_history.history['val_loss'][ft_best]:.3f} (epoch {ft_best + 1})")
```

**Observations from fine-tuning (last 40 layers, lr 5e-5)**

- Trainable parameters rose to about 2M and validation accuracy to about 0.85.
- Validation loss stopped improving after about 6 epochs while training accuracy kept rising, so overfitting increased.

```python
import time

# redefining fine_tune with a verbose option so the experiments dont print 15 lines of logs each
# (same as before, default verbose=1 behaves the same, and it returns the history now too)
def fine_tune(model, base_model, n_unfreeze=40, lr=5e-5, epochs=15, name='cnn_finetune', verbose=1):
    base_model.trainable = True
    for layer in base_model.layers[:-n_unfreeze]:
        layer.trainable = False
    # keep the batchnorm layers frozen
    for layer in base_model.layers:
        if isinstance(layer, tf.keras.layers.BatchNormalization):
            layer.trainable = False

    model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=lr),
                  loss='categorical_crossentropy', metrics=['accuracy'])

    cbs = [
        tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=4, restore_best_weights=True),
        tf.keras.callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=2, min_lr=1e-7),
        tf.keras.callbacks.ModelCheckpoint(f"{MODEL_DIR}/{name}.keras",
                                           monitor='val_loss', save_best_only=True)
    ]
    hist = model.fit(train_prep, validation_data=val_prep, epochs=epochs,
                     class_weight=class_weights, callbacks=cbs, verbose=verbose)
    return hist

# one experiment = build a fresh model, train the new layers, then fine tune
def run_experiment(name, dropout_rate=0.3, dense_units=256, l2_reg=1e-4, n_unfreeze=40, ft_lr=5e-5):
    start = time.time()
    tf.keras.utils.set_random_seed(42)   # same seed so the runs are comparable
    m, base = build_model(dropout_rate=dropout_rate, dense_units=dense_units, l2_reg=l2_reg)
    m = compile_model(m, learning_rate=1e-3)

    # step 1: train only the new layers (base frozen)
    m.fit(train_prep, validation_data=val_prep, epochs=10, class_weight=class_weights,
          callbacks=[tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=3,
                                                      restore_best_weights=True)],
          verbose=0)

    # step 2: fine tune the top layers
    h = fine_tune(m, base, n_unfreeze=n_unfreeze, lr=ft_lr, epochs=15, name=name, verbose=0)

    # results at the epoch with the lowest validation loss
    best = int(np.argmin(h.history['val_loss']))
    res = {
        'name': name, 'dropout': dropout_rate, 'l2': l2_reg, 'dense_units': dense_units,
        'unfrozen_layers': n_unfreeze, 'ft_lr': ft_lr, 'best_epoch': best + 1,
        'train_acc': h.history['accuracy'][best],
        'val_acc': h.history['val_accuracy'][best],
        'val_loss': h.history['val_loss'][best],
    }
    res['gap'] = res['train_acc'] - res['val_acc']
    print(f"{name}: val acc {res['val_acc']:.3f}, val loss {res['val_loss']:.3f}, "
          f"gap {res['gap']:.3f}, best epoch {res['best_epoch']} ({(time.time() - start) / 60:.1f} min)")
    return res

# changing one thing at a time compared to the baseline
experiments = [
    dict(name='exp_baseline'),
    dict(name='exp_more_reg', dropout_rate=0.5, l2_reg=1e-3),
    dict(name='exp_unfreeze_80', n_unfreeze=80),
    dict(name='exp_lr_1e4', ft_lr=1e-4),
]

experiment_results = []
for cfg in experiments:
    experiment_results.append(run_experiment(**cfg))

# results table, saved to drive in case colab disconnects
exp_df = pd.DataFrame(experiment_results).round(3)
print("\nExperiment results (chosen on validation data only):")
print(exp_df)
exp_df.to_csv(f"{MODEL_DIR}/cnn_experiments.csv", index=False)
```

**Observations from the tuning experiments**

- One change at a time compared with the baseline, judged on validation data only.
- Unfreezing 80 layers was best (highest validation accuracy, lowest loss). A higher learning rate overfit the most, and more regularization did not help.
- Differences under about 2% are within run-to-run noise on this validation set.
- The depth and width of the custom layers were not tuned because of time.

```python
# Final CNN: retrain the best configuration from round 1 (last 80 layers unfrozen)
# on the fixed split, so early stopping and checkpointing only see the new validation set
# and the test set is used once, here
final_res = run_experiment('final_cnn_unfreeze80', n_unfreeze=80)

# fine_tune saved the checkpoint with the lowest validation loss, loading that one
final_cnn = tf.keras.models.load_model(f"{MODEL_DIR}/final_cnn_unfreeze80.keras")

# single evaluation on the clean test set, same function as the earlier models
final_results = evaluate_cnn(final_cnn, test_prep, 'Final CNN (MobileNetV2, last 80 layers fine-tuned)')
show_errors(final_results)
```

**Final CNN results and error analysis**

### 3.3 Train and Evaluate the Model

```python
# TODO: Train the text classification model
# - Use appropriate training parameters
# - Monitor training progress

import copy
import time
from torch.utils.data import DataLoader

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
train_loader = DataLoader(train_text_ds, batch_size=32, shuffle=True)
val_loader = DataLoader(val_text_ds, batch_size=64)
test_loader = DataLoader(test_text_ds, batch_size=64)

# loss and accuracy on a dataset without training (dropout off)
def evaluate_text(model, loader):
    model.eval()
    total_loss, preds, labels = 0.0, [], []
    with torch.no_grad():
        for batch in loader:
            batch = {k: v.to(device) for k, v in batch.items()}
            out = model(**batch)
            total_loss += out.loss.item() * len(batch['labels'])
            preds.extend(out.logits.argmax(-1).cpu().numpy())
            labels.extend(batch['labels'].cpu().numpy())
    return total_loss / len(labels), accuracy_score(labels, preds)

# one experiment: fresh model, adamw with weight decay, linear warmup then decay,
# early stopping on validation loss and keeping the best epoch's weights
def train_text_model(name, lr=5e-5, epochs=5, weight_decay=0.01, freeze_base=False, patience=2):
    start = time.time()
    torch.manual_seed(42); np.random.seed(42); random.seed(42)
    model = build_text_model().to(device)
    if freeze_base:   # only the classification head learns
        for p in model.distilbert.parameters():
            p.requires_grad = False

    params = [p for p in model.parameters() if p.requires_grad]
    optimizer = torch.optim.AdamW(params, lr=lr, weight_decay=weight_decay)
    total_steps = epochs * len(train_loader)
    warmup = max(1, int(0.1 * total_steps))
    scheduler = torch.optim.lr_scheduler.LambdaLR(
        optimizer, lambda s: min((s + 1) / warmup, max(0.0, (total_steps - s) / (total_steps - warmup))))

    history = {'train_loss': [], 'train_acc': [], 'val_loss': [], 'val_acc': []}
    best_loss, best_state, best_epoch, bad_epochs = float('inf'), None, 0, 0
    for epoch in range(epochs):
        model.train()
        run_loss, correct, seen = 0.0, 0, 0
        for batch in train_loader:
            batch = {k: v.to(device) for k, v in batch.items()}
            out = model(**batch)
            optimizer.zero_grad()
            out.loss.backward()
            optimizer.step()
            scheduler.step()
            run_loss += out.loss.item() * len(batch['labels'])
            correct += (out.logits.argmax(-1) == batch['labels']).sum().item()
            seen += len(batch['labels'])

        val_loss, val_acc = evaluate_text(model, val_loader)
        history['train_loss'].append(run_loss / seen)
        history['train_acc'].append(correct / seen)
        history['val_loss'].append(val_loss)
        history['val_acc'].append(val_acc)
        print(f"  {name} epoch {epoch + 1}: train loss {run_loss / seen:.4f} acc {correct / seen:.4f} | "
              f"val loss {val_loss:.4f} acc {val_acc:.4f}")

        if val_loss < best_loss:
            best_loss, best_epoch, bad_epochs = val_loss, epoch + 1, 0
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
        else:
            bad_epochs += 1
            if bad_epochs >= patience:
                print(f"  early stopping after epoch {epoch + 1}")
                break

    model.load_state_dict(best_state)
    model.to('cpu')   # park it on the cpu so several runs fit in memory
    result = {'name': name, 'lr': lr, 'freeze_base': freeze_base, 'best_epoch': best_epoch,
              'val_loss': best_loss, 'val_acc': history['val_acc'][best_epoch - 1],
              'minutes': round((time.time() - start) / 60, 1)}
    return model, history, result

# experiments, chosen on validation data only
# accuracy is close to 1.0 for most settings, so validation loss is used to pick the winner
# (lower loss = the model is confidently right, not just barely right)
text_experiments = [
    dict(name='frozen_base', lr=1e-3, freeze_base=True),
    dict(name='full_lr_2e-5', lr=2e-5),
    dict(name='full_lr_5e-5', lr=5e-5),
    dict(name='full_lr_1e-4', lr=1e-4),
]
text_runs = {}
for cfg in text_experiments:
    print(cfg['name'])
    m, h, r = train_text_model(**cfg)
    text_runs[cfg['name']] = (m, h, r)

text_exp_df = pd.DataFrame([r for _, _, r in text_runs.values()]).round({'val_loss': 4, 'val_acc': 4})
print("\nDistilBERT experiments (validation):")
print(text_exp_df)

# pick the best run and plot its curves
best_text_name = text_exp_df.loc[text_exp_df['val_loss'].idxmin(), 'name']
text_model, text_history, _ = text_runs[best_text_name]
text_model.to(device)
print("Chosen DistilBERT run:", best_text_name)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
for key, ax, title in [('loss', axes[0], 'Loss'), ('acc', axes[1], 'Accuracy')]:
    ax.plot(range(1, len(text_history[f'train_{key}']) + 1), text_history[f'train_{key}'], label='train')
    ax.plot(range(1, len(text_history[f'val_{key}']) + 1), text_history[f'val_{key}'], label='validation')
    ax.set_title(f'{best_text_name}: {title}')
    ax.set_xlabel('Epoch')
    ax.legend()
plt.tight_layout()
plt.show()

# save the chosen model to drive for part 5
TEXT_MODEL_DIR = f"{DRIVE_DIR}/models/distilbert_waste"
text_model.save_pretrained(TEXT_MODEL_DIR)
tokenizer.save_pretrained(TEXT_MODEL_DIR)
print("Saved to", TEXT_MODEL_DIR)
```

- No overfitting: training and validation loss fell together. Techniques: AdamW with weight decay, warmup then decay, early stopping.

### 3.4 Create Classification Function

```python
# TODO: Create a function that takes a text description and returns the predicted waste category

def classify_waste_description(description, return_details=False):
    """
    Classifies a waste description into an appropriate category.

    Args:
        description (str): Text description of waste item
        return_details (bool): if True, also return confidence and the top 3 categories
                               (used by the integrated assistant in part 5)

    Returns:
        str: Predicted waste category
        (or a dict with category, confidence, top3 and clean_text if return_details=True)
    """
    # input checks so the assistant gets a clear error instead of a crash
    if not isinstance(description, str) or not description.strip():
        raise ValueError("Please provide a non-empty text description.")
    text = clean_text(description)          # same cleaning as the training data
    if not text:
        raise ValueError("The description has no words after cleaning (only numbers or symbols).")

    probs = bert_predict_proba([text])[0]
    order = probs.argsort()[::-1]
    category = str(le.classes_[order[0]])
    if not return_details:
        return category
    return {
        'category': category,
        'confidence': round(float(probs[order[0]]), 3),
        'top3': [(str(le.classes_[i]), round(float(probs[i]), 3)) for i in order[:3]],
        'clean_text': text,
    }

# quick checks: normal input, messy input, a low-confidence case and invalid inputs
print(classify_waste_description("Empty Coca-Cola can!!"))
print(classify_waste_description("  OLD Newspaper, slightly wet  "))
print(classify_waste_description("leftover ugali and sukuma wiki", return_details=True))
for bad_input in ["", "   ", "123 !!!", None]:
    try:
        classify_waste_description(bad_input)
    except ValueError as e:
        print(f"{bad_input!r} -> error: {e}")
```

## Part 4: Recycling Instruction Generation with RAG

In this section, you will implement a Retrieval-Augmented Generation (RAG) system to generate recycling instructions.

### 4.1 Preprocess Documents for Retrieval

```python
# TODO: Prepare documents for retrieval
# - Process policy documents and disposal instructions
# - Create embeddings for efficient retrieval

!pip install -q faiss-cpu sentence-transformers
import re
import faiss
from sentence_transformers import SentenceTransformer

# rag_df (37 chunks with category, source, priority) was built in section 1.3

# --- 1. factual consistency: the documents contradict each other (section 1.2)
# decision: the category policy wins, so any line in a combined policy that the category
# policy lists as Non-Acceptable is removed before indexing
def parse_sections(text):
    # headings are lines ending with ':' (both "Non-Acceptable Items:" and "GLASS GUIDELINES:")
    sections, current = {}, None
    for line in text.split('\n'):
        line = line.strip()
        if not line:
            continue
        if line.endswith(':') and not line.startswith('-'):
            current = line[:-1].strip()
            sections[current] = []
        elif current is not None:
            sections[current].append(line)
    return sections

def norm_item(line):
    return re.sub(r'\s+', ' ', line.lstrip('-').strip().lower())

# non-acceptable items per category, taken from the category policies (policies 1-9)
not_accepted = {}
for doc in policy_docs:
    if len(doc['categories_covered']) == 1:
        items = parse_sections(doc['document_text']).get('Non-Acceptable Items', [])
        not_accepted[doc['categories_covered'][0]] = {norm_item(x) for x in items}

removed_lines = []
def drop_conflicts(row):
    if row['source'] != 'combined_policy':
        return row['text']
    banned = not_accepted.get(row['category'], set())
    kept = []
    for line in row['text'].split('\n'):
        if line.startswith('-') and norm_item(line) in banned:
            removed_lines.append({'policy_id': int(row['policy_id']), 'category': row['category'], 'line': line})
        else:
            kept.append(line)
    return '\n'.join(kept)

if 'text_original' not in rag_df.columns:      # so rerunning the cell does not filter twice
    rag_df['text_original'] = rag_df['text']
rag_df['text'] = rag_df.apply(lambda r: drop_conflicts({**r, 'text': r['text_original']}), axis=1)
print("Contradicting lines removed from combined policies:")
print(pd.DataFrame(removed_lines))

# --- 2. embeddings: comparing two small sentence-embedding models
# all-MiniLM-L6-v2: fast general-purpose model, 384 dimensions
# bge-small-en-v1.5: trained specifically for retrieval, 384 dimensions, uses a prefix on queries
EMBED_MODELS = {
    'all-MiniLM-L6-v2': ('sentence-transformers/all-MiniLM-L6-v2', ''),
    'bge-small-en-v1.5': ('BAAI/bge-small-en-v1.5', 'Represent this sentence for searching relevant passages: '),
}

# evaluation queries with the category whose documents should come back:
# the required format "How to recycle {category}" plus item-level questions a resident might ask
retrieval_queries = [(f"How to recycle {c}", c) for c in all_categories] + [
    ("can I put a greasy pizza box in recycling", 'Cardboard'),
    ("what do I do with empty cereal boxes", 'Cardboard'),
    ("where do eggshells and coffee grounds go", 'Food Organics'),
    ("can I compost meat and fish bones", 'Food Organics'),
    ("are broken mirrors recyclable", 'Glass'),
    ("do I need to remove lids from jam jars", 'Glass'),
    ("can I recycle empty aerosol cans", 'Metal'),
    ("what about old paint cans", 'Metal'),
    ("where do I take dead batteries", 'Miscellaneous Trash'),
    ("how do I get rid of old medication", 'Miscellaneous Trash'),
    ("which plastic codes are accepted", 'Plastic'),
    ("can black plastic containers be recycled", 'Plastic'),
    ("can I recycle shredded paper", 'Paper'),
    ("are receipts recyclable", 'Paper'),
    ("what do I do with old clothes and towels", 'Textile Trash'),
    ("can wet or moldy fabric be recycled", 'Textile Trash'),
    ("how long can tree branches be for yard waste", 'Vegetation'),
    ("can I put soil and rocks in the garden waste bin", 'Vegetation'),
]
q_texts = [q for q, _ in retrieval_queries]
q_cats = [c for _, c in retrieval_queries]
chunk_cats = rag_df['category'].values

def build_index(vectors):
    # vectors are normalized, so inner product = cosine similarity
    index = faiss.IndexFlatIP(vectors.shape[1])
    index.add(vectors.astype('float32'))
    return index

embedders, indexes, retrieval_rows = {}, {}, []
for name, (path, q_prefix) in EMBED_MODELS.items():
    model_e = SentenceTransformer(path, device=str(device))
    doc_vecs = model_e.encode(rag_df['text'].tolist(), normalize_embeddings=True)
    index = build_index(doc_vecs)
    q_vecs = model_e.encode([q_prefix + q for q in q_texts], normalize_embeddings=True)
    _, top = index.search(q_vecs.astype('float32'), 3)
    # semantic search only (no category filter): does the right category come back?
    top1 = np.mean([chunk_cats[t[0]] == c for t, c in zip(top, q_cats)])
    top3 = np.mean([np.mean(chunk_cats[t] == c) for t, c in zip(top, q_cats)])
    retrieval_rows.append({'embedding_model': name, 'top1_category_acc': round(top1, 3),
                           'top3_category_precision': round(top3, 3)})
    embedders[name] = (model_e, q_prefix)
    indexes[name] = (index, top)

retrieval_eval_df = pd.DataFrame(retrieval_rows)
print("\nSemantic search only, no category filter (27 queries):")
print(retrieval_eval_df)

# keep the better model
EMBED_NAME = retrieval_eval_df.loc[retrieval_eval_df['top1_category_acc'].idxmax(), 'embedding_model']
embedder, query_prefix = embedders[EMBED_NAME]
rag_index, best_top = indexes[EMBED_NAME]
print("Chosen embedding model:", EMBED_NAME)

# queries where pure semantic search pulled in the wrong category
print("\nSemantic-only misses with the chosen model:")
for (q, c), t in zip(retrieval_queries, best_top):
    got = list(chunk_cats[t])
    if got[0] != c:
        print(f"  '{q}' (expected {c}) -> top 3: {got}")

# --- 3. retrieval function used by the generator
# hybrid retrieval: when the category is known (from the cnn or text classifier) only that
# category's chunks are searched, ranked by priority first (category policy and disposal
# instructions before combined policies) and then by similarity to the query
def retrieve(query, category=None, k=3):
    q_vec = embedder.encode([query_prefix + query], normalize_embeddings=True).astype('float32')
    scores, ids = rag_index.search(q_vec, len(rag_df))   # score every chunk (only 37)
    results = []
    for score, i in zip(scores[0], ids[0]):
        row = rag_df.iloc[i]
        if category is not None and row['category'] != category:
            continue
        results.append({'chunk_id': row['chunk_id'], 'category': row['category'], 'source': row['source'],
                        'policy_id': None if pd.isna(row['policy_id']) else int(row['policy_id']),
                        'priority': int(row['priority']), 'score': round(float(score), 3),
                        'text': row['text']})
    if category is not None:
        results.sort(key=lambda r: (r['priority'], -r['score']))
    return results[:k]

print("\nExample: retrieve('How to recycle Glass', category='Glass')")
for r in retrieve("How to recycle Glass", category='Glass'):
    print(f"  {r['chunk_id']} | {r['source']} | policy {r['policy_id']} | priority {r['priority']} | score {r['score']}")
```

**Observations from document preparation and retrieval**

- 4 contradicting lines were removed from the combined policies before indexing (window glass and plastic bags).
- bge-small-en-v1.5 put the right category first more often than all-MiniLM-L6-v2 (about 0.93 vs 0.85), so it was chosen.
- Pure semantic search still mixed categories (a soil question retrieved Food Organics), so retrieval filters by the predicted category, then ranks by priority and similarity.
- FAISS exact cosine search, instant for 37 chunks.

### 4.2 Implement RAG-based System

```python
# TODO: Select a pre-trained language model and implement RAG
# - Choose an appropriate language model
# - Create a retrieval mechanism

from transformers import AutoModelForSeq2SeqLM

# generator: flan-t5-base (about 250M parameters)
# why: it is instruction-tuned, so it already follows prompts like "answer using the policy below",
# and its encoder-decoder design suits reading a context and writing a new text from it.
# it is small enough to fine-tune on a t4 in minutes. decoder-only models like gpt-2 are not
# instruction-tuned and tend to continue the text instead of answering from the documents
GEN_MODEL_NAME = 'google/flan-t5-base'
gen_tokenizer = AutoTokenizer.from_pretrained(GEN_MODEL_NAME)
gen_model = AutoModelForSeq2SeqLM.from_pretrained(GEN_MODEL_NAME).to(device)
print(f"Generator parameters: {sum(p.numel() for p in gen_model.parameters()):,}")

GEN_MAX_INPUT = 512   # t5 input limit used for training and generation

# retrieval step of RAG: the category's top chunks joined into one context
def build_context(category, query=None, k=3):
    query = query or f"How to recycle {category}"
    chunks = retrieve(query, category=category, k=k)
    context = "\n\n".join(c['text'] for c in chunks)
    return context, chunks

# prompt that tells the model to answer from the retrieved policy only
def build_prompt(category, context, query=None):
    query = query or f"How to recycle {category}?"
    return (f"Answer the question using only the Metro City recycling policy below. "
            f"Say what is accepted, what is not accepted, how to prepare the items and how they are collected.\n\n"
            f"Policy:\n{context}\n\nQuestion: {query}\nAnswer:")

def generate_text(prompt, model=None, max_new_tokens=200, **gen_kwargs):
    model = model if model is not None else gen_model
    model.eval()
    enc = gen_tokenizer(prompt, return_tensors='pt', truncation=True, max_length=GEN_MAX_INPUT).to(device)
    with torch.no_grad():
        out = model.generate(**enc, max_new_tokens=max_new_tokens, **gen_kwargs)
    return gen_tokenizer.decode(out[0], skip_special_tokens=True)

# prompt length per category: must stay under 512 tokens or the question at the end gets cut off
prompt_lengths = {}
for cat in all_categories:
    ctx, _ = build_context(cat)
    prompt_lengths[cat] = len(gen_tokenizer(build_prompt(cat, ctx))['input_ids'])
print("Prompt length in tokens:", prompt_lengths)
print("Longest prompt:", max(prompt_lengths.values()), "tokens (limit", GEN_MAX_INPUT, ")")

# zero-shot baseline: the pretrained model with retrieved context, no fine-tuning, greedy decoding
zero_shot_outputs = {}
for cat in all_categories:
    ctx, _ = build_context(cat)
    zero_shot_outputs[cat] = generate_text(build_prompt(cat, ctx))

for cat in ['Glass', 'Plastic', 'Food Organics', 'Miscellaneous Trash']:
    print(f"\n--- {cat} (zero-shot) ---")
    print(zero_shot_outputs[cat])
```

### 4.3 Adjust and Evaluate the System

```python
# TODO: Train the RAG-based system
# - Adjust sampling methods/parameters

# --- 1. training data: answers in a fixed format built from the category policy
SECTION_LABELS = {
    'Acceptable Items': 'Accepted',
    'Non-Acceptable Items': 'Not accepted',
    'Items That Cannot Be Recycled': 'Cannot be recycled',
    'Special Handling Items': 'Special handling',
    'Preparation Instructions': 'Preparation',
    'Collection Method': 'Collection',
}

def make_answer(policy_text):
    sections = parse_sections(policy_text)
    parts = []
    for heading, label in SECTION_LABELS.items():   # same section order for every category
        lines = sections.get(heading, [])
        if lines:
            items = '; '.join(l.lstrip('-').strip().rstrip('.') for l in lines)
            parts.append(f"{label}: {items}.")
    return ' '.join(parts)

# augmentation: randomly drop list items from the policy in the context and build the target
# from the same reduced policy, so the model must copy from the context instead of memorizing
def drop_items(policy_text, rng, keep_prob=0.75):
    kept_lines, current = [], None
    for line in policy_text.split('\n'):
        s = line.strip()
        if s.endswith(':') and not s.startswith('-'):
            current = s[:-1]
        if s.startswith('-') and current in SECTION_LABELS and rng.random() > keep_prob:
            continue
        kept_lines.append(line)
    return '\n'.join(kept_lines)

QUERY_TEMPLATES = ["How to recycle {}?", "How do I recycle {}?",
                   "How should I dispose of {}?", "What are the rules for recycling {}?"]

def make_examples(n_per_cat, seed):
    rng = random.Random(seed)
    examples = []
    for cat in all_categories:
        base_chunks = retrieve(f"How to recycle {cat}", category=cat, k=3)
        for _ in range(n_per_cat):
            texts, policy_text = [], None
            for ch in base_chunks:
                if ch['source'] == 'category_policy':
                    policy_text = drop_items(ch['text'], rng)
                    texts.append(policy_text)
                else:
                    texts.append(ch['text'])
            query = rng.choice(QUERY_TEMPLATES).format(cat)
            examples.append({'category': cat,
                             'prompt': build_prompt(cat, '\n\n'.join(texts), query),
                             'target': make_answer(policy_text)})
    return examples

gen_train = make_examples(40, seed=1)
gen_val = make_examples(5, seed=2)
print("Training examples:", len(gen_train), "| validation examples:", len(gen_val))
print("Example target:", gen_train[0]['target'][:200], "...")

# --- 2. fine-tune flan-t5 (fresh copy, so rerunning the cell does not train twice)
gen_model = AutoModelForSeq2SeqLM.from_pretrained(GEN_MODEL_NAME).to(device)

class GenDataset(torch.utils.data.Dataset):
    def __init__(self, examples):
        self.enc = gen_tokenizer([e['prompt'] for e in examples], truncation=True,
                                 max_length=GEN_MAX_INPUT, padding='max_length', return_tensors='pt')
        labels = gen_tokenizer([e['target'] for e in examples], truncation=True,
                               max_length=256, padding='max_length', return_tensors='pt')['input_ids']
        labels[labels == gen_tokenizer.pad_token_id] = -100   # padding is ignored in the loss
        self.labels = labels
    def __len__(self):
        return len(self.labels)
    def __getitem__(self, i):
        return {'input_ids': self.enc['input_ids'][i],
                'attention_mask': self.enc['attention_mask'][i],
                'labels': self.labels[i]}

def gen_val_loss(model, loader):
    model.eval()
    total, n = 0.0, 0
    with torch.no_grad():
        for b in loader:
            b = {k: v.to(device) for k, v in b.items()}
            total += model(**b).loss.item() * len(b['labels'])
            n += len(b['labels'])
    return total / n

torch.manual_seed(42)
gen_train_loader = DataLoader(GenDataset(gen_train), batch_size=4, shuffle=True)
gen_val_loader = DataLoader(GenDataset(gen_val), batch_size=8)
optimizer = torch.optim.AdamW(gen_model.parameters(), lr=3e-4, weight_decay=0.01)

GEN_EPOCHS = 4
best_gen_loss, best_gen_state = float('inf'), None
start = time.time()
for epoch in range(GEN_EPOCHS):
    gen_model.train()
    run_loss = 0.0
    for b in gen_train_loader:
        b = {k: v.to(device) for k, v in b.items()}
        loss = gen_model(**b).loss
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        run_loss += loss.item() * len(b['labels'])
    val_loss = gen_val_loss(gen_model, gen_val_loader)
    print(f"epoch {epoch + 1}: train loss {run_loss / len(gen_train):.4f} | val loss {val_loss:.4f}")
    if val_loss < best_gen_loss:   # keep the best epoch
        best_gen_loss = val_loss
        best_gen_state = {k: v.detach().cpu().clone() for k, v in gen_model.state_dict().items()}
gen_model.load_state_dict(best_gen_state)
print(f"best val loss {best_gen_loss:.4f}, training took {(time.time() - start) / 60:.1f} min")

# save to drive for part 5
GEN_MODEL_DIR = f"{DRIVE_DIR}/models/flan_t5_recycling"
gen_model.save_pretrained(GEN_MODEL_DIR)
gen_tokenizer.save_pretrained(GEN_MODEL_DIR)
print("Saved to", GEN_MODEL_DIR)

# --- 3. sampling methods on one category (the full comparison is in the evaluation cell)
SAMPLING_SETTINGS = {
    'greedy': dict(do_sample=False),
    'beam_4': dict(do_sample=False, num_beams=4),
    'top_p_0.9_temp_0.7': dict(do_sample=True, top_p=0.9, temperature=0.7),
    'top_k_50_temp_1.2': dict(do_sample=True, top_k=50, temperature=1.2),
}
glass_ctx, _ = build_context('Glass')
for name, params in SAMPLING_SETTINGS.items():
    torch.manual_seed(42)
    print(f"\n--- Glass, {name} ---")
    print(generate_text(build_prompt('Glass', glass_ctx), max_new_tokens=256, **params))
```

**Observations from fine-tuning**

- 360 training / 45 validation examples. About 25% of policy items are randomly removed from the context and the target, so the model must read the context instead of memorizing.
- Fixed answer format: Accepted / Not accepted / Preparation / Collection.
- Validation loss reached about 0 within 2 to 3 epochs (AdamW, lr 3e-4, 4 epochs, best epoch kept).
- The fine-tuned Glass answer covers the whole policy; the zero-shot answers were one sentence, and Food Organics was wrong.

```python
# TODO: Evaluate the quality of generated instructions
# - Test with various waste categories
# - Assess relevance and accuracy

!pip install -q textstat
import textstat

LABEL_PATTERN = r'(Accepted|Not accepted|Cannot be recycled|Special handling|Preparation|Collection):'

# split a generated answer into its sections and items
def parse_answer(text):
    parts = re.split(LABEL_PATTERN, text)
    sections = {}
    for label, body in zip(parts[1::2], parts[2::2]):
        items = [x.strip().rstrip('.').strip() for x in body.split(';')]
        sections[label] = [x for x in items if x]
    return sections

def flat(text):
    return re.sub(r'\s+', ' ', text.lower().replace('\n- ', ' ').replace('\n', ' ')).strip()

policy_text_for = {r['category']: r['text'] for _, r in rag_df[rag_df['source'] == 'category_policy'].iterrows()}

def score_output(cat, text, context):
    # reference = every item in the full category policy
    ref_items = {norm_item(i) for its in parse_answer(make_answer(policy_text_for[cat])).values() for i in its}
    sections = parse_answer(text)
    if sections:
        items = [i for its in sections.values() for i in its]
    else:   # zero-shot answers have no labels, so they are scored sentence by sentence
        items = [s.strip().rstrip('.') for s in re.split(r'(?<=\.)\s+', text) if s.strip()]
    out_items = {norm_item(i) for i in items}
    ctx = flat(context)
    return {
        'coverage': np.mean([i in out_items for i in ref_items]),                     # policy items included
        'grounded': np.mean([norm_item(i) in ctx for i in items]) if items else 0.0,   # items found in the documents
        'contradictions': sum(norm_item(i) in not_accepted.get(cat, set())             # banned items listed as accepted
                              for i in sections.get('Accepted', [])),
        'flesch': textstat.flesch_reading_ease(text),                                 # higher = easier to read
        'words': len(text.split()),
    }

# --- 1. zero-shot vs fine-tuned with different sampling settings, all 9 categories
eval_contexts = {cat: build_context(cat)[0] for cat in all_categories}
EVAL_SETTINGS = {**SAMPLING_SETTINGS, 'top_k_50_temp_2.0': dict(do_sample=True, top_k=50, temperature=2.0)}

rows, ft_outputs = [], {}
for cat in all_categories:
    rows.append({'setting': 'zero-shot greedy', 'category': cat,
                 **score_output(cat, zero_shot_outputs[cat], eval_contexts[cat])})
for name, params in EVAL_SETTINGS.items():
    for cat in all_categories:
        torch.manual_seed(42)
        out = generate_text(build_prompt(cat, eval_contexts[cat]), max_new_tokens=256, **params)
        ft_outputs[(name, cat)] = out
        rows.append({'setting': f'fine-tuned {name}', 'category': cat, **score_output(cat, out, eval_contexts[cat])})

gen_eval_df = pd.DataFrame(rows)
print("Average over the 9 categories:")
print(gen_eval_df.groupby('setting', sort=False)[['coverage', 'grounded', 'contradictions', 'flesch', 'words']]
      .mean().round(3))
print("\nFine-tuned greedy, per category:")
print(gen_eval_df[gen_eval_df['setting'] == 'fine-tuned greedy'].round(3).to_string(index=False))
print("\nExample at temperature 2.0 (Plastic):")
print(ft_outputs[('top_k_50_temp_2.0', 'Plastic')])

# --- 2. does the model really read the documents? edit the policy in the context:
# add a made-up accepted item and delete one non-accepted item, then check the answer follows
edit_tests = [('Glass', 'Glass marbles', 'Light bulbs'),
              ('Metal', 'Copper pipes', 'Pressurized containers'),
              ('Paper', 'Paper egg trays', 'Thermal receipt paper')]
print("\nContext edit test:")
for cat, new_item, removed_item in edit_tests:
    edited = re.sub(r'(?m)^Acceptable Items:$', f'Acceptable Items:\n- {new_item}', policy_text_for[cat], count=1)
    edited = edited.replace(f'- {removed_item}\n', '')
    others = [c['text'] for c in build_context(cat)[1] if c['source'] != 'category_policy']
    out = generate_text(build_prompt(cat, '\n\n'.join([edited] + others)), max_new_tokens=256)
    out_sections = parse_answer(out)
    all_out = {norm_item(i) for its in out_sections.values() for i in its}
    print(f"  {cat}: added '{new_item}' listed as accepted: "
          f"{norm_item(new_item) in {norm_item(i) for i in out_sections.get('Accepted', [])}} | "
          f"removed '{removed_item}' gone from answer: {norm_item(removed_item) not in all_out}")
```

**Observations from evaluation**

- Zero-shot: almost no policy coverage (about 0.01). Fine-tuned: full coverage, everything grounded in the documents, 0 contradictions.
- Sampling: all settings gave the same answer except temperature 2.0, which drifted from the policy wording. Greedy is used: deterministic, fast and faithful.
- Context edit test: when an item was added to or removed from the policy, the answer followed in 3 of 3 cases, so the model relies on the retrieved documents.
- Readability is low (Flesch about 32) because each section is one long sentence. The 4.4 function formats answers as bullet points.
- Limitation: the model mostly copies policy lines, and the checks use exact text matching.

### 4.4 Create Instruction Generation Function

```python
# TODO: Create a function that takes a waste category and generates recycling instructions

# turn the generated answer into readable bullet points
def format_instructions(category, sections):
    lines = [f"How to recycle {category} (Metro City policy)"]
    for label, items in sections.items():
        lines.append(f"\n{label}:")
        lines += [f"- {i}" for i in items]
    return '\n'.join(lines)

def generate_recycling_instructions(waste_category, query=None, k=3, **gen_kwargs):
    """
    Generates detailed recycling instructions for a given waste category.

    Args:
        waste_category (str): Waste category
        query (str): optional resident question, defaults to "How to recycle {category}?"
        k (int): number of policy chunks to retrieve
        **gen_kwargs: optional decoding settings (default is greedy decoding)

    Returns:
        str: Detailed recycling instructions
        list: Relevant policy documents
    """
    # accept any capitalisation, reject unknown categories with a clear message
    if not isinstance(waste_category, str) or not waste_category.strip():
        raise ValueError("Please provide a waste category.")
    match = [c for c in all_categories if c.lower() == waste_category.strip().lower()]
    if not match:
        raise ValueError(f"Unknown category '{waste_category}'. Valid categories: {', '.join(all_categories)}")
    category = match[0]

    # retrieve, then generate from the retrieved policy
    context, chunks = build_context(category, query=query, k=k)
    answer = generate_text(build_prompt(category, context, query), max_new_tokens=256, **gen_kwargs)

    sections = parse_answer(answer)
    if not sections:   # fallback if the model ever ignores the format
        instructions = answer.strip()
    else:
        # safety check: never list an item as accepted if the category policy says it is not
        banned = not_accepted.get(category, set())
        if 'Accepted' in sections:
            sections['Accepted'] = [i for i in sections['Accepted'] if norm_item(i) not in banned]
        instructions = format_instructions(category, sections)

    documents = [{'source': c['source'], 'policy_id': c['policy_id'],
                  'title': c['text'].split('\n')[0], 'text': c['text']} for c in chunks]
    return instructions, documents

# example
instructions, docs = generate_recycling_instructions('glass')
print(instructions)
print("\nRelevant policy documents:")
for d in docs:
    print(f"  - {d['title']} ({d['source']}, policy {d['policy_id']})")

# readability before vs after formatting (each bullet line counted as its own sentence)
def as_sentences(text):
    return ' '.join(l.strip().rstrip('.') + '.' for l in text.split('\n') if l.strip())

read_rows = []
for cat in all_categories:
    raw = ft_outputs[('greedy', cat)]
    formatted = format_instructions(cat, parse_answer(raw))
    read_rows.append({'category': cat,
                      'flesch_raw': textstat.flesch_reading_ease(raw),
                      'flesch_bullets': textstat.flesch_reading_ease(as_sentences(formatted))})
read_df = pd.DataFrame(read_rows).round(1)
print("\nReadability (Flesch reading ease, higher = easier):")
print(read_df.to_string(index=False))
print("Average:", read_df[['flesch_raw', 'flesch_bullets']].mean().round(1).to_dict())

# invalid inputs
for bad in ['Electronics', '', None]:
    try:
        generate_recycling_instructions(bad)
    except ValueError as e:
        print(f"{bad!r} -> error: {e}")
```

**Observations from the instruction function**

- Steps: check the category (any capitalisation, clear error otherwise), retrieve the top 3 chunks, generate with greedy decoding, remove any "accepted" item the category policy bans (safety net), format as bullets, return the instructions and the source documents.
- Bullet formatting raised the Flesch score from about 32 to about 41 on average. Miscellaneous Trash is still hard to read (about 12) because of long technical words.

### 5.1 Design Integration Architecture

```python
# TODO: Design an architecture that integrates all three models
# - Create interfaces between components
# - Handle input/output flow

from PIL import Image

# cnn: use the one in memory, otherwise load the saved final model from drive
if 'final_cnn' not in globals():
    final_cnn = tf.keras.models.load_model(f"{DRIVE_DIR}/models/final_cnn_unfreeze80.keras")
# text_model (distilbert, part 3) and gen_model (flan-t5, part 4) are already loaded

# image interface: accepts a file path, a PIL image or a numpy array
def load_input_image(image):
    if isinstance(image, (str, pathlib.Path)):
        if not os.path.exists(image):
            raise FileNotFoundError(f"Image not found: {image}")
        try:
            image = Image.open(image)
        except Exception:
            raise ValueError(f"Could not read the file as an image: {image}")
    if isinstance(image, Image.Image):
        image = np.array(image.convert('RGB'))
    image = np.asarray(image)
    if image.ndim != 3 or image.shape[-1] not in (3, 4):
        raise ValueError("Image must be a colour image (height x width x 3).")
    image = image[..., :3].astype('float32')
    # same steps as training: bilinear resize to 224x224, then mobilenet normalization
    image = tf.image.resize(image, (IMG_HEIGHT, IMG_WIDTH), method='bilinear').numpy()
    return preprocess_input(image[np.newaxis, ...])

def classify_image(image):
    probs = final_cnn.predict(load_input_image(image), verbose=0)[0]
    order = probs.argsort()[::-1]
    return {'category': class_names[order[0]],
            'confidence': round(float(probs[order[0]]), 3),
            'top3': [(class_names[i], round(float(probs[i]), 3)) for i in order[:3]]}

# choosing the image confidence threshold on the validation set
val_probs = final_cnn.predict(val_prep, verbose=0)
val_true = np.concatenate([y.numpy().argmax(axis=1) for _, y in val_prep])
val_conf, val_pred = val_probs.max(axis=1), val_probs.argmax(axis=1)
print("Image confidence thresholds (validation set):")
for t in [0.5, 0.6, 0.7, 0.8]:
    keep = val_conf >= t
    print(f"  {t}: answers {keep.mean():.0%} of images | accuracy when answered {(val_pred[keep] == val_true[keep]).mean():.3f}"
          f" | accuracy of flagged ones {(val_pred[~keep] == val_true[~keep]).mean():.3f}")

# text threshold from the part 3 challenge set (all predictions below 0.85 were wrong)
TEXT_CONF_THRESHOLD = 0.85

# quick check of both interfaces
print("\nImage:", test_files[0], "->", classify_image(test_files[0]))
print("Text: 'empty coke can' ->", classify_waste_description("empty coke can", return_details=True))
```

**Architecture and design choices**

- Flow: input (image or text) → matching classifier (MobileNetV2 CNN or DistilBERT) → confidence check → category → RAG (bge-small retrieval filtered by category + fine-tuned flan-t5) → instructions and source documents.
- Both classifiers use the same 9 category names, so the predicted category passes straight to the RAG step.
- Image input accepts a file path, a PIL image or an array, and is preprocessed exactly as in training (bilinear resize to 224x224, MobileNetV2 normalization).
- Confidence thresholds: 0.7 for images (on the validation set it answers about 89% of images with about 0.93 accuracy, and below 0.7 the CNN is right less than half the time) and 0.85 for text (from the Part 3 challenge set).

### 5.2 Implement Integrated Assistant

```python
# TODO: Implement the integrated waste management assistant

import time
from functools import lru_cache

IMAGE_CONF_THRESHOLD = 0.7   # chosen on the validation set in 5.1

# instructions only depend on the category, so each one is generated once and reused (faster)
@lru_cache(maxsize=None)
def cached_instructions(category):
    return generate_recycling_instructions(category)

history = []   # every request, used for feedback

def waste_management_assistant(input_data, input_type="image"):
    """
    Integrated waste management assistant that processes either images or text descriptions
    and returns waste classification and recycling instructions.

    Args:
        input_data: Either an image file path/array or a text description
        input_type (str): Type of input - "image" or "text"

    Returns:
        dict: Dictionary containing waste category, confidence, and recycling instructions
    """
    start = time.time()
    try:
        if input_type == 'image':
            pred, threshold = classify_image(input_data), IMAGE_CONF_THRESHOLD
        elif input_type == 'text':
            pred, threshold = classify_waste_description(input_data, return_details=True), TEXT_CONF_THRESHOLD
        else:
            raise ValueError("input_type must be 'image' or 'text'.")
        instructions, docs = cached_instructions(pred['category'])
    except Exception as e:   # bad input or a model problem returns a clear error instead of crashing
        return {'status': 'error', 'message': str(e)}

    low = pred['confidence'] < threshold
    history.append({'input_type': input_type, 'predicted': pred['category'], 'confidence': pred['confidence']})
    return {
        'request_id': len(history) - 1,
        'status': 'low_confidence' if low else 'ok',
        'category': pred['category'],
        'confidence': pred['confidence'],
        'alternatives': [c for c, _ in pred['top3'][1:]],
        'message': (f"Not sure ({pred['confidence']:.0%}). Please confirm the category or add a description."
                    if low else f"This looks like {pred['category']}."),
        'instructions': instructions,
        'policy_documents': docs,
        'seconds': round(time.time() - start, 2),
    }

# feedback: the user corrects the category, the correction is saved to drive for the next
# retraining, and the user gets the instructions for the right category straight away
FEEDBACK_PATH = f"{DRIVE_DIR}/feedback_log.csv"
if os.path.exists(FEEDBACK_PATH):
    os.remove(FEEDBACK_PATH)   # fresh log on every notebook run (demo); a real system would keep it

def give_feedback(request_id, correct_category):
    if correct_category not in class_names:
        raise ValueError(f"Valid categories: {', '.join(class_names)}")
    entry = {**history[request_id], 'correct_category': correct_category}
    pd.DataFrame([entry]).to_csv(FEEDBACK_PATH, mode='a', header=not os.path.exists(FEEDBACK_PATH), index=False)
    return cached_instructions(correct_category)[0]

# quick demo: one test image, then a low-confidence description corrected through feedback
r = waste_management_assistant(test_files[0], 'image')
print(r['status'], '|', r['message'], f"({r['seconds']} s)")
print(r['instructions'])
print("Sources:", [d['title'] for d in r['policy_documents']])

r = waste_management_assistant("leftover ugali and sukuma wiki", 'text')
print('\n' + r['status'], '|', r['message'], '| alternatives:', r['alternatives'])
print("After the user corrects it to Food Organics:")
print(give_feedback(r['request_id'], 'Food Organics'))
```

**Observations from the assistant**

- One function handles both input types: classify → confidence check → instructions and source documents. Any bad input or model error returns a clear error message instead of crashing.
- Below the confidence threshold the status is "low_confidence": the user still sees the best guess and its alternatives, and is asked to confirm or describe the item.
- Instructions are cached per category: the first request takes about 2 to 3 seconds (generation), and later requests for the same category are almost instant.
- Feedback: give_feedback() saves the correction to a log on Drive (for the next retraining) and immediately returns the instructions for the correct category.

### 5.3 Evaluate the Integrated System

```python
# TODO: Evaluate the integrated system on test cases
# - Test with images from test dataset
# - Test with text descriptions from test dataset
# - Assess overall performance

# run many inputs through the full assistant
def run_batch(inputs, true_labels, input_type):
    rows = []
    for x, y in zip(inputs, true_labels):
        r = waste_management_assistant(x, input_type)
        rows.append({'request_id': r.get('request_id'), 'true': y, 'pred': r.get('category'),
                     'status': r['status'], 'seconds': r.get('seconds')})
    df = pd.DataFrame(rows)
    df['correct'] = df['true'] == df['pred']
    return df

# --- 1. end-to-end tests: all test images, all test descriptions, and the part 3 challenge set
results = {
    'images (test set)': run_batch(test_files, [pathlib.Path(p).parent.name for p in test_files], 'image'),
    'text (test set)': run_batch(test_df['description'], test_df['category'], 'text'),
    'text (challenge set)': run_batch(ch_df['description'], ch_df['category'], 'text'),
}
summary = []
for name, df in results.items():
    ok = df['status'] == 'ok'
    summary.append({'test': name, 'n': len(df), 'accuracy': df['correct'].mean(),
                    'flagged_low_conf': (~ok).mean(),
                    'acc_when_confident': df.loc[ok, 'correct'].mean(),
                    'acc_when_flagged': df.loc[~ok, 'correct'].mean() if (~ok).any() else np.nan,
                    'avg_seconds': df['seconds'].mean()})
print(pd.DataFrame(summary).round(3).to_string(index=False))

# --- 2. edge cases
edge_cases = [
    ('missing file', 'no_such_image.jpg', 'image'),
    ('not an image', 'waste_descriptions.csv', 'image'),
    ('grayscale image', Image.open(test_files[0]).convert('L'), 'image'),
    ('tiny random image', np.random.randint(0, 255, (8, 8, 3)), 'image'),
    ('empty text', '', 'text'),
    ('only numbers', '12345 !!!', 'text'),
    ('very long text', 'old glass jar ' * 200, 'text'),
    ('wrong input type', 'glass jar', 'audio'),
    ('image path as text', test_files[0], 'text'),
]
print("\nEdge cases:")
for name, x, kind in edge_cases:
    r = waste_management_assistant(x, kind)
    print(f"  {name:20} -> {r['status']}: {r['message']}")

# --- 3. feedback loop: correct the challenge-set mistakes as a user would
wrong = results['text (challenge set)'].query('not correct')
for _, row in wrong.iterrows():
    give_feedback(int(row['request_id']), row['true'])
fb = pd.read_csv(FEEDBACK_PATH)
print(f"\nFeedback log: {len(fb)} entries. Most common corrections (predicted -> correct):")
print(fb.groupby(['predicted', 'correct_category']).size().sort_values(ascending=False).head(5))
```

**Observations from testing the integrated system**

- Images: about 0.90 end-to-end. 11% were flagged, and flagged answers were much less accurate (about 0.62 vs 0.93).
- Text: about 0.999 on the test set. On the challenge set the threshold caught all 6 uncertain errors.
- Fast: about 0.13 s per image and 0.01 s per text request, thanks to caching.
- Edge cases (missing or non-image files, empty text, wrong input type) return clear errors. Grayscale images and very long text still work.
- Limitations: nonsense inputs still get a category. Feedback shows the text model falls back to Miscellaneous Trash or Food Organics for unfamiliar items.

## Conclusion and recommendations

**What was built**
- An assistant that takes an image or a text description, predicts one of 9 waste categories, and returns Metro City recycling instructions together with the policy documents they come from.
- Image: MobileNetV2 fine-tuned on RealWaste. Text: fine-tuned DistilBERT. Instructions: retrieval with bge-small and FAISS, generated by a fine-tuned flan-t5-base.

**Key results**
- CNN: about 0.89 test accuracy. The hardest cases are Miscellaneous Trash and the look-alike pairs Metal/Plastic and Glass/Plastic.
- Text: about 0.999 on the test set. On 36 realistic descriptions DistilBERT scored about 0.75 against 0.64 for TF-IDF, because it copes better with unseen words.
- Instructions: full policy coverage, everything grounded in the documents, 0 contradictions. The answer changed when the policy text changed (3 of 3), so the model relies on the retrieved documents.
- Assistant: about 0.90 end-to-end on test images. Low-confidence answers are flagged, and these were much less accurate, so the flag works.

**Key decisions**
- Fixed the provided image split, which leaked images between validation and test.
- Category policies win over combined policies when they conflict. Contradicting lines are removed before indexing.
- Confidence thresholds were chosen from data (0.7 for images, 0.85 for text) rather than set by hand.

**Limitations**
- The text data is generated and templated, so test scores overstate real-world performance.
- All photos have similar grey backgrounds, so real resident photos may score lower.
- The generator mostly copies policy lines instead of writing new explanations.
- Some errors are made with high confidence, and nonsense inputs still receive a category.
- The challenge set is small (36 items) and was written by us.

**Next steps**
- Test with real resident photos and descriptions.
- Add a check that rejects inputs that are not waste items.
- Use the feedback log to retrain on corrected examples, and have people review the generated instructions for clarity.
